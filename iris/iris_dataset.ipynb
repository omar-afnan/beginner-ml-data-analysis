**1 target column: species.**

```python
import pandas as pd
df = pd.read_csv("/content/iris_dataset.csv")
print
```

```python
# species

df["species"].value_counts()  #flowers tht belong to each species.
```

```python
df.groupby('species').mean(numeric_only=True)
# find the avg measurements
```

```python
import matplotlib.pyplot as plt
df.groupby("species")["petal_length"].mean().plot(kind="bar")
plt.title("avg petal lenth by the species ")
plt.xlabel("flower species ")
plt.ylabel("avg petal length")
plt.show()
```

```python
# scatter plot
import matplotlib.pyplot as plt

for species in df["species"].unique():
    subset = df[df["species"] == species]

    plt.scatter(
        subset["petal_length"],
        subset["petal_width"],
        label=species
    )

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Petal Length vs Petal Width")
plt.legend()
plt.show()
```

```python
versicolor_data = df[df["species"] == 'versicolor']
versicolor_data
```

```python
# findinf the mean of the petal lenth and width for versicolor_data
versi_color_length = versicolor_data['petal_length'].mean()
versi_color_width = versicolor_data["petal_width"].mean()

print(versi_color_length)
print(versi_color_width)
```

```python
setosa_data = df[df["species"] == 'setosa']
setosa_data
```

```python
# Calculate the mean petal length and petal width for setosa
mean_petal_length = setosa_data['petal_length'].mean()
mean_petal_width = setosa_data['petal_width'].mean()

print(f"Setosa Petal Length Mean: {mean_petal_length:.3f}")
print(f"Setosa Petal Width Mean: {mean_petal_width:.3f}")
```

```python
# virginica
virginica_data = df[df["species"] == 'virginica']
virginica_data
```

```python
# now finding it petal lenghr and width
virginica_length = virginica_data["petal_length"].mean()
virginica_width = virginica_data["petal_width"].mean()

print(virginica_length)
print(virginica_width)
```

```python
correlation = df.corr(numeric_only=True)

print(correlation)
```

```python
# heat map

import seaborn as sns
import matplotlib.pyplot as plt


sns.heatmap(
    correlation,
    annot = True,
    cmap = "coolwarm",
    fmt = ".2f" # how many decimal to place

)


plt.title("Iris Feature Correlation")
plt.show()
```
